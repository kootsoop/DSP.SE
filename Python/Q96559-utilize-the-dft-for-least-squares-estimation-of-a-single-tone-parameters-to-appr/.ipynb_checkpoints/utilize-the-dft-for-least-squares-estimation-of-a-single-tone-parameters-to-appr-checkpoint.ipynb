```python
import numpy as np
import csv
import matplotlib.pyplot as plt

data = []
with open('array.csv', mode ='r')as file:
  csvFile = csv.reader(file)
  for lines in csvFile:
        data.append(float(lines[0]))

plt.plot(data[1:300])
```

```python

plt.plot(data)
plt.plot(np.abs(np.fft.fft(data)))
```

```python
import numpy as np
def loss_function(x,a,phi,omega):
    t = np.linspace(0,len(x),len(x))
    return np.sum(np.square((a*np.cos(omega*t+phi)-x)))
```

```python
ahat = np.sum(np.abs(data))/len(data)
```

```python
error = []
omega_range = np.linspace(0,np.pi)
for omega in omega_range:
    error.append(loss_function(data,2,omega,np.pi/4))
plt.plot(omega_range, error)
ix = np.argmin(error)
plt.plot(omega_range[ix], error[ix], 'ro')
print(omega_range[ix]/2/np.pi)
```

```python
error = ahat*np.cos(omega_range[ix]*np.linspace(1,len(data),len(data)))-data
```

```python
plt.plot(error)
```

```python
import scipy

def loss_again(omega, x):
    return loss_function(x,2,omega,np.pi/4)

print(scipy.optimize.least_squares(loss_again, omega_range[ix]))
```
