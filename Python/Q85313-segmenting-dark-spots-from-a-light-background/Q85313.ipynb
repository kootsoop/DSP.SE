```python
import os
from PIL import Image
import numpy as np

os.chdir('/Volumes/Macintosh HD 1/Users/kootsoop/git/DSP.SE/Python/Q85313-segmenting-dark-spots-from-a-light-background')

def rgb_to_gray_threshold(img):
	blue_yellow_image = np.zeros(len(img.getdata()))
	for i, pixel in enumerate(img.getdata()):
		r, g, b = pixel
		if b > 60 and r < 100 and g < 100:
			blue_yellow_image[i] = 255
		else:
			blue_yellow_image[i] = 0

	return blue_yellow_image

def rgb_to_gray_tim(img):
	blue_yellow_image = np.zeros(len(img.getdata()))
	for i, pixel in enumerate(img.getdata()):
		r, g, b = pixel
		blue_yellow_image[i] = int((b - (r + g) / 2) + 128)

	return blue_yellow_image

img = Image.open("images/shadow1.jpeg")
img.show()

blue_yellow_image_tim = rgb_to_gray_tim(img)
new_im_tim = Image.fromarray(blue_yellow_image_tim.reshape(img.size[1], img.size[0]).astype(np.uint8))
new_im_tim.show()
new_im_tim.save("images/blue_yellow_image_shadow1_tim.jpeg")

blue_yellow_image_threshold = rgb_to_gray_threshold(img)
new_im_threshold = Image.fromarray(blue_yellow_image_threshold.reshape(img.size[1], img.size[0]).astype(np.uint8))
new_im_threshold.show()
new_im_threshold.save("images/blue_yellow_image_shadow1_threshold.jpeg")
```
