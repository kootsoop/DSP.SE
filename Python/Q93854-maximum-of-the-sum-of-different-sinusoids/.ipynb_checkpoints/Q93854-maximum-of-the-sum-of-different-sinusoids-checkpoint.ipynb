```python
import numpy as np
from matplotlib import pyplot as plt

def calculate_signal(A, omega, phi, t):
    signal = np.zeros(len(t))
    for index in range(len(A)):
        signal = signal + A*sin(omega*t + phi)
        
    return signal

def calculate_signal_derivative(A, omega, phi, t):
    signal = np.zeros(len(t))
    for index in range(len(A)):
        signal = signal + A*omega*cos(omega*t + phi)
        
    return signal


# Sinusoidal parameters
A = [1,2,3,4]
omega = [ np.random.uniform(0, 2 * np.pi) for x in A]
phi = [np.random.uniform(0, 2 * np.pi) for x in A]

# Where to start
t_peak_guess = np.random.uniform(0,100)

# 

```
