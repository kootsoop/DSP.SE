```python
import numpy as np
from matplotlib import pyplot as plt

def calculate_signal(A, omega, phi, t):
    if isinstance(t, float):
        signal = 0
    else:
        signal = np.zeros(len(t))
    for index in range(len(A)):
        signal = signal + A[index]*np.sin(omega[index]*t + phi[index])
        
    return signal

def calculate_signal_derivative(A, omega, phi, t):
    if isinstance(t, float):
        signal = 0
    else:
        signal = np.zeros(len(t))
    for index in range(len(A)):
        signal = signal + A[index]*omega[index]*np.cos(omega[index]*t + phi[index])
        
    return signal

def calculate_signal_second_derivative(A, omega, phi, t):
    if isinstance(t, float):
        signal = 0
    else:
        signal = np.zeros(len(t))
    for index in range(len(A)):
        signal = signal + -A[index]*omega[index]*omega[index]*np.cos(omega[index]*t + phi[index])
        
    return signal


# Sinusoidal parameters
A = [1,2,3,4]
omega = [ np.random.uniform(0, np.pi/4) for x in A]
phi = [np.random.uniform(0, np.pi*2) for x in A]

tt = np.linspace(0,100,10000)
plt.plot(tt,calculate_signal(A,omega,phi,tt))


def find_peak(A,omega,phi):
    # Where to start
    t_peak_guess = np.random.uniform(0,100)
    x_peak_guess = calculate_signal(A,omega,phi, t_peak_guess)

    plt.plot(t_peak_guess, x_peak_guess, 'g.')

    # Newton iteration
    for iter in range(10):
        t_peak_guess = t_peak_guess - calculate_signal_derivative(A,omega,phi, t_peak_guess)/calculate_signal_second_derivative(A,omega,phi, t_peak_guess)
        x_peak_guess = calculate_signal(A,omega,phi, t_peak_guess)
        plt.plot(t_peak_guess, x_peak_guess, 'rx')

    return t_peak_guess, x_peak_guess


t, x = find_peak(A,omega,phi)
plt.plot(t, x, 'ko')

t, x = find_peak(A,omega,phi)
plt.plot(t, x, 'ko')

t, x = find_peak(A,omega,phi)
plt.plot(t, x, 'ko')

t, x = find_peak(A,omega,phi)
plt.plot(t, x, 'ko')

```
